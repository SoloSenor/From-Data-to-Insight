# فصل چهارم: نگهداری و سازمان‌دهی داده در پایتون

دفترچهٔ همراه متن فصل؛ توضیح کامل، تمرین و خطاهای عمدی در فایل متن قرار دارند. هر سلول کد مستقل است. از هستهٔ تازه، همهٔ سلول‌ها را به‌ترتیب اجرا کنید. تمام کدها و خروجی‌ها انگلیسی‌اند؛ توضیحات فارسی بیرون سلول کد قرار می‌گیرند. C22 عمداً خطای منطقی دارد؛ C23 مقدار مشتق‌شدهٔ قدیمی را نشان می‌دهد. C24 فقط سه رکورد مشخص را پردازش می‌کند و ابزار گزارش عمومی نیست.

## مثال C01 — ساخت، طول و دسترسی

```python
quantities = [3, 1, 4, 2]
print(quantities)
print(len(quantities))
print(quantities[0])
print(quantities[-1])
```

## مثال C02 — برش‌های پایه

```python
quantities = [3, 1, 4, 2]
print(quantities[1:3])
print(quantities[:2])
print(quantities[2:])
print(quantities[::2])
print(quantities[::-1])
print(quantities)
```

## مثال C03 — مرز برش و فهرست خالی

```python
quantities = [3, 1, 4, 2]
print(quantities[:99])
print(quantities[99:])
print(quantities[2:2])
print([][:2])
```

## مثال C04 — جایگزینی مقدار موجود

```python
quantities = [3, 1, 4]
quantities[1] = 2
print(quantities)
```

## مثال C05 — یک عنصر یا چند عنصر؟

```python
first = [3, 1]
first.append([4, 2])
second = [3, 1]
second.extend([4, 2])
print(first)
print(len(first))
print(second)
print(len(second))
```

## مثال C06 — درج و برداشتن

```python
order_ids = ["O-0401", "O-0403"]
order_ids.insert(1, "O-0402")
removed_id = order_ids.pop()
print(order_ids)
print(removed_id)
```

## مثال C07 — حذف با مقدار و حذف با موقعیت

```python
labels = ["new", "paid", "new", "sent"]
labels.remove("new")
print(labels)
del labels[1]
print(labels)
```

## مثال C08 — وجود، رخداد و طول

```python
categories = ["book", "pen", "book"]
print("book" in categories)
print("bag" in categories)
print(categories.count("book"))
print(len(categories))
```

## مثال C09 — مرتب‌سازی بدون تغییر و با تغییر

```python
amounts_toman = [720_000, 160_000, 300_000]
ascending = sorted(amounts_toman)
print(ascending)
print(amounts_toman)
result = amounts_toman.sort(reverse=True)
print(amounts_toman)
print(result)
```

## مثال C10 — دو نام برای یک فهرست

## مثال C19 — تعداد رخدادها در برابر تعداد دسته‌ها

```python
categories = ["book", "pen", "book", "bag"]
unique_categories = set(categories)
print(len(categories))
print(len(unique_categories))
print(sorted(unique_categories))
print("pen" in unique_categories)
```

## مثال C20 — عملیات مجموعه و افزودن عضو

```python
expected_ids = {"O-0401", "O-0402", "O-0403"}
received_ids = {"O-0401", "O-0403", "O-0404"}
print(sorted(expected_ids & received_ids))
print(sorted(expected_ids - received_ids))
print(sorted(received_ids - expected_ids))
print(sorted(expected_ids | received_ids))
received_ids.add("O-0404")
print(len(received_ids))
```

## مثال C21 — فهرستی از رکوردها

```python
orders = [
    {"order_id": "O-0401", "quantity": 3, "unit_price_toman": 240_000},
    {"order_id": "O-0402", "quantity": 2, "unit_price_toman": 80_000},
]
print(len(orders))
print(orders[0]["order_id"])
print(orders[1]["quantity"])
print(orders[1]["quantity"] * orders[1]["unit_price_toman"])
```

## مثال C22 — خروجی ظاهراً معتبر با معنای نادرست

هشدار: این نمونه عمداً ارتباط شناسه و مبلغ را خراب می‌کند؛ خروجی، گزارش درست نیست.

```python
order_ids = ["O-0401", "O-0402", "O-0403"]
amounts_toman = [720_000, 160_000, 300_000]
amounts_toman.sort()
print(order_ids[0], amounts_toman[0])
print(len(order_ids) == len(amounts_toman))
```

## مثال C23 — مقدار مشتق‌شده، رابطهٔ زنده نیست

هشدار: مبلغ ذخیره‌شده با تغییر تعداد خودکار به‌روز نمی‌شود.

```python
order = {"quantity": 3, "unit_price_toman": 240_000}
order["gross_toman"] = order["quantity"] * order["unit_price_toman"]
order["quantity"] = 2
print(order["gross_toman"])
print(order["quantity"] * order["unit_price_toman"])
```

## مثال C24 — ساخت رکوردها و گزارش قابل‌کنترل

پیش‌شرط: سه سفارش تک‌قلمی، بازپرداخت کامل هر واحد، بدون سایر تعدیل‌ها. مبلغ پس از مرجوعی سود نیست. افزودن رکورد چهارم در محاسبه پوشش داده نمی‌شود.

```python
orders = [
    {
        "order_id": "O-0401",
        "quantity": 3,
        "unit_price_toman": 240_000,
        "returned_quantity": 1,
    },
    {
        "order_id": "O-0402",
        "quantity": 2,
        "unit_price_toman": 80_000,
        "returned_quantity": 0,
    },
    {
        "order_id": "O-0403",
        "quantity": 1,
        "unit_price_toman": 300_000,
        "returned_quantity": 1,
    },
]

first = orders[0]
second = orders[1]
third = orders[2]

gross_amounts_toman = [
    first["quantity"] * first["unit_price_toman"],
    second["quantity"] * second["unit_price_toman"],
    third["quantity"] * third["unit_price_toman"],
]
refund_amounts_toman = [
    first["returned_quantity"] * first["unit_price_toman"],
    second["returned_quantity"] * second["unit_price_toman"],
    third["returned_quantity"] * third["unit_price_toman"],
]

gross_total_toman = sum(gross_amounts_toman)
refund_total_toman = sum(refund_amounts_toman)
after_returns_toman = gross_total_toman - refund_total_toman

print(f"Order records: {len(orders)}")
print(f"Gross amounts: {gross_amounts_toman}")
print(f"Refund amounts: {refund_amounts_toman}")
print(f"Gross total (toman): {gross_total_toman:,}")
print(f"Refund total (toman): {refund_total_toman:,}")
print(f"After returns (toman): {after_returns_toman:,}")
```
