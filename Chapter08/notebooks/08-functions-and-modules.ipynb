## 01. تعریف، فراخوانی و مقدار بازگشتی

تعریف تابع دستور محاسبه را آماده می‌کند؛ فقط فراخوانی آن محاسبه را انجام می‌دهد.

در تعریف، quantity و unit_price_toman پارامترند؛ در فراخوانی، 2 و 120000 آرگومان‌اند. result نتیجهٔ فراخوانی را می‌گیرد. بدنۀ تابع با چهار فاصله تورفته است. return نتیجه را به فراخواننده تحویل می‌دهد، نه اینکه آن را الزاماً روی صفحه نمایش دهد.

```python
def amount(quantity, unit_price_toman):
    return quantity * unit_price_toman

result = amount(2, 120000)
print(result)
```

## 03. بازگرداندن چند نتیجه در یک تاپل

سه مبلغ را با یک فراخوانی محاسبه می‌کنیم.

تابع در واقع یک شیء تاپل برمی‌گرداند و سمت چپ آن را باز می‌کند. ترتیب سه مؤلفه بخشی از قرارداد است. برای نتیجه‌های بسیار زیاد یا نام‌محور، دیکشنری می‌تواند روشن‌تر باشد. مقدار باقی‌مانده سود نیست؛ هزینه‌ها در این سناریو وجود ندارند.

```python
def amounts(quantity, unit_price_toman, returned_quantity=0):
    gross = quantity * unit_price_toman
    refund = returned_quantity * unit_price_toman
    return gross, refund, gross - refund

gross, refund, retained = amounts(2, 450000, 1)
print(gross, refund, retained)
```

## 06. پارامتر فقط کلیدی

علامت ستارهٔ تنها، پارامترهای پس از خود را فقط کلیدی می‌کند.

ستاره در این تعریف نه ضرب است و نه جمع‌آوری آرگومان؛ مرز قراردادیِ نحوهٔ فراخوانی است. فراخوانی meets_minimum(450000, 500000) خطای TypeError دارد. این الگو برای گزینه‌های نمایش که ممکن است با عدد اصلی اشتباه شوند مفید است.

```python
def meets_minimum(value, *, minimum=0):
    return value >= minimum

print(meets_minimum(450000, minimum=500000))
print(meets_minimum(690000, minimum=500000))
```

## 11. دامنهٔ محلی و هم‌نامی

نام یکسان در دو دامنه الزاماً به یک متغیر محلی مشترک اشاره نمی‌کند.

انتساب label در تابع، نام محلی می‌سازد و اتصال بیرونی را تغییر نمی‌دهد. این موضوع با تغییر محتوای یک فهرست مشترک متفاوت است. برای تغییر اتصال سراسری باید صریحاً global به کار رود؛ این فصل به‌جای تکیه بر آن، ورودی و خروجی روشن را توصیه می‌کند.

```python
label = "global"
def local_label():
    label = "local"
    return label

print(local_label())
print(label)
```

## 12. دامنهٔ دربرگیرنده

جست‌وجوی نام فقط محلی و سراسری نیست؛ تابع تودرتو می‌تواند نام دامنهٔ بیرونی خود را بخواند.

در توابع معمول، الگوی ذهنی LEGB یعنی محلی، دربرگیرنده، سراسریِ ماژول و داخلی پایتون. این خلاصه همهٔ جزئیات کلاس‌ها یا مدل‌های پیشرفتهٔ دامنه نیست. برای خواندن نام لازم نیست global یا nonlocal بنویسیم؛ این دو برای تغییر اتصال نام در دامنهٔ دیگر مطرح می‌شوند.

```python
def outer():
    label = "enclosing"
    def inner():
        return label
    return inner()

print(outer())
```

## 15. تغییر ظرف ورودی در برابر ساخت نتیجهٔ تازه

تفاوت دو تابع در اثر جانبی است، نه صرفاً شکل return.

تابع نخست ورودی را تغییر می‌دهد و None برمی‌گرداند؛ تابع دوم فهرست بیرونی تازه می‌سازد. برای عضوهای تو‌در‌تو، فهرست تازه به معنی کپی عمیق نیست. تغییر درجا می‌تواند انتخاب درست و کارآمدی باشد، به شرط نام‌گذاری، مستندسازی و انتظار روشن فراخواننده.

```python
def add_in_place(values, value):
    values.append(value)

def added(values, value):
    return values + [value]

source = [10]
print(add_in_place(source, 20))
print(source)
print(added(source, 30))
print(source)
```

## 16. دام پیش‌فرض تغییرپذیر

این مثال عمداً رفتار نامطلوب دارد ولی استثنا نمی‌دهد.

فهرست پیش‌فرض هنگام اجرای def ساخته شده و میان فراخوانی‌ها مشترک است. اینجا def یک بار اجرا شده است؛ هر اجرای دوبارهٔ تعریف، پیش‌فرض تازه‌ای می‌سازد و ممکن است عیب را در دفترچه پنهان کند. برای بررسی، دو فراخوانی پی‌درپی پس از یک تعریف لازم است.

```python
def remember_bad(value, items=[]):
    items.append(value)
    return items

print(remember_bad("A"))
print(remember_bad("B"))
```

## 17. پیش‌فرض امن و قرارداد مالکیت

در این نسخه، حتی فهرستِ داده‌شده از بیرون هم تغییر نمی‌کند.

الگوی None فقط مشکل پیش‌فرض مشترک را حل می‌کند؛ حفظ فهرستِ صریح بیرونی حاصل copy در این نسخه است. اگر items را مستقیم تغییر می‌دادیم، فراخواننده هنوز تغییر را می‌دید. شرط is None دقیق‌تر از if not items است، چون فهرست خالیِ صریح را با نبود آرگومان یکی نمی‌کند.

```python
def remembered(value, items=None):
    result = [] if items is None else items.copy()
    result.append(value)
    return result

print(remembered("A"))
print(remembered("B"))
source = ["old"]
print(remembered("new", source))
print(source)
```

## 18. زمان ارزیابی پیش‌فرض

پیش‌فرض معمولاً با هر فراخوانی از روی مقدار فعلی یک متغیر دوباره محاسبه نمی‌شود.

ارزیابی هنگام اجرای تعریف انجام شده است. این نکته برای تاریخ جاری، عدد تصادفی، اتصال منبع و تنظیماتِ تغییرپذیر مهم است. برای محاسبهٔ تازه، ورودی را صریح بفرستید یا مقدار مناسب را در بدنۀ تابع بسازید؛ پیش‌فرضِ فراخوانی یک تابع هم هنگام تعریف ارزیابی می‌شود.

```python
limit = 10
def configured_limit(value=limit):
    return value

limit = 20
print(configured_limit())
print(configured_limit(limit))
```

## 19. مستندسازی قرارداد با docstring

نخستین رشتهٔ بدنۀ تابع مستند آن است و ابزارها می‌توانند آن را بخوانند.

مستند فقط بازگویی نام تابع نیست: واحد، پیش‌شرط، خروجی، اثر جانبی و محدودیت باید روشن باشند. راهنمای مفصل فارسی در متن کتاب قرار می‌گیرد؛ docstring داخل کد طبق قرارداد کتاب انگلیسی است. help نیز راهنما را نمایش می‌دهد، اما خروجی طولانی و وابسته به محیط آن را جزو خروجی مرجع نکرده‌ایم.

```python
def net_units(quantity, returned_quantity=0):
    """Return retained units for a valid single-item order."""
    return quantity - returned_quantity

print(net_units.__doc__)
print(net_units(5, 1))
```

## 26. خواندن امضای تابع

## 30. قرارداد خروجیِ فیلتر رکوردها

تابع را به یک نام سراسریِ خاص وابسته نمی‌کنیم؛ داده و گزینه را می‌گیرد.

هر دیکشنری انتخاب‌شده کم‌عمق کپی شده است و تغییر فیلد سطح اول نمونه، اصل را عوض نمی‌کند. تابع نه روی صفحه چاپ می‌کند و نه دادهٔ سراسری را می‌خواند. اگر رکورد دارای فیلد تو‌در‌تو بود، این قرارداد به معنی استقلال عمیق آن فیلد نمی‌بود.

```python
def select_category(rows, category):
    return [row.copy() for row in rows if row["category"] == category]

books = select_category(orders, "Books")
books[0]["quantity"] = 99
print(len(books))
print(orders[0]["quantity"])
```
