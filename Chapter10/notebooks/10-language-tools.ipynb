## C01 — فضای نامِ روشن با import
فصل هشتم سازوکار import را معرفی کرد. اینجا مسئلهٔ ما انتخاب ابزار و خوانایی وابستگی است: نام ماژول کمک می‌کند منشأ تابع را ببینیم.

```python
import math
print(math.isqrt(81))
print(math.isclose(0.1 + 0.2, 0.3))
```

تابع isqrt ریشهٔ صحیح را به‌صورت عدد صحیح می‌دهد؛ تابع isclose برابری دقیق نیست و معیار نزدیکی دارد. واردکردن math به‌معنای چسباندن متن فایل ماژول به برنامه نیست. برای آزمایش عددهای نزدیک صفر، معمولاً باید abs_tol متناسب با واحد و مسئله تعیین شود؛ پیش‌فرض نسبی به‌تنهایی کافی نیست. استفاده از نام ماژول، وابستگی را جلوی چشم خواننده نگه می‌دارد.

## C02 — واردکردن ابزار مشخص از statistics
اگر معنای نام‌ها روشن و تعداد ابزارها کم است، واردکردن نام مشخص نیز مناسب است. ابزار درست را پس از فهم سؤال انتخاب می‌کنیم.

```python
from statistics import mean, median
values = [10, 12, 14, 100]
print(mean(values))
print(median(values))
```

این اختلاف، اثر مشاهدهٔ بزرگ بر میانگین را نشان می‌دهد، نه اینکه میانه همیشه بهتر است. واحد هر دو همان واحد داده است. این مثال کوچک، آموزش استنباط آماری نیست. تابع mean جای بررسی مقدار گمشده، واحد مخلوط یا نمونه‌گیری نامناسب را نمی‌گیرد. نام mean را بعداً با یک عدد بازنویسی نکنید؛ در آن صورت دیگر نام تابع در آن فضای نام نیست.

## C03 — تاریخ به‌عنوان شیء، نه رشتهٔ ظاهراً مرتب
تاریخ معتبر رفتار تقویمی دارد. کم‌کردن دو رشتهٔ شبیه تاریخ تعریف نشده است، اما دو شیء date اختلاف روزها را می‌دانند.

```python
from datetime import date, timedelta
start = date.fromisoformat("2024-02-28")
end = date.fromisoformat("2024-03-01")
print((end - start).days)
print((start + timedelta(days=2)).isoformat())
```

دو روز بودن فاصله به سال کبیسه مربوط است. date تاریخ میلادیِ بدون ساعت و منطقهٔ زمانی است، نه تاریخ شمسی و نه یک لحظهٔ زمانی جهانی. تبدیل تقویم، timezone و تاریخ‌های مبهم موضوع این نمونه نیستند. fromisoformat در Python 3.11 چند شکل ISO را می‌پذیرد؛ برای قرارداد دقیق YYYY-MM-DD همان کنترل بازتولید isoformat فصل نهم لازم است.

## C04 — Decimal و قرارداد نمایش عدد
برای بعضی محاسبه‌های ده‌دهی، Decimal انتخاب آگاهانه‌تری است. ورودی متنی باعث می‌شود اعشار موردنظر را مستقیماً وارد کنیم.

```python
from decimal import Decimal
print(0.1 + 0.2 == 0.3)
print(Decimal("0.1") + Decimal("0.2"))
print(Decimal("12.345").quantize(Decimal("0.01")))
```

گردکردن خط آخر در زمینهٔ پیش‌فرض این اجرا ROUND_HALF_EVEN است؛ سیاست مالی باید صریحاً تعیین شود. Decimal ساخته‌شده از float، تقریب دودویی موجود را وارد می‌کند. Decimal هم دقت زمینه و قواعد گردکردن دارد و همهٔ مسائل عددی را خودکار حل نمی‌کند. در پروژهٔ سفارش‌ها قیمت صحیح به تومان است و برای همان قرارداد به Decimal نیاز نداریم. ابزار تازه نباید واحد داده را بی‌دلیل عوض کند.

## C05 — تکرارپذیر و دو تکرارگر مستقل
فهرست را می‌توان دوباره پیمود. هر فراخوانی iter روی این فهرست یک وضعیت پیمایش مستقل می‌سازد؛ خود داده‌ها الزاماً کپی نمی‌شوند.

```python
values = [10, 20, 30]
left = iter(values)
right = iter(values)
print(next(left), next(left), next(right))
print(list(left))
print(values)
```

left دو گام جلو رفته و right فقط یک گام. list(left) باقی‌ماندهٔ همان پیمایش را مصرف می‌کند، نه اینکه از اول فهرست شروع کند. وضعیت پیمایش را از وضعیت ظرف جدا کنید. تغییر ظرف هنگام پیمایش ممکن است نتیجه را تغییر دهد یا در بعضی ظرف‌ها خطا بدهد؛ مستقل‌بودن تکرارگر معادل snapshot مستقل از داده نیست.

## C06 — تکرارگر تک‌مصرف است
یک تابع می‌تواند روی فهرست درست کار کند و روی تکرارگر نتیجهٔ ناقص بدهد. مسئله در اینجا خطای نحوی نیست؛ مصرف ورودی است.

```python
source = iter([2, 3, 5])
print(iter(source) is source)
print(sum(source))
print(sum(source))
```

در قرارداد iterator، iter همان تکرارگر را برمی‌گرداند. جمع دوم صفر است چون چیزی باقی نمانده، نه چون مجموع اصلی صفر بوده است. برای دو شاخص از یک ورودی تک‌مصرف، هر دو را در یک حلقه حساب کنید، یا با تصمیم آگاهانه ورودی را materialize کنید. تبدیل به list هزینهٔ حافظه دارد و باید قبل از تمام‌شدن داده انجام شود.

## C07 — range مولد نیست
کم‌حافظه بودن یک شیء به‌تنهایی آن را generator نمی‌کند. طبقه‌بندی دقیق به ما می‌گوید چه عملیاتی انتظار داشته باشیم.

```python
values = range(2, 8, 2)
print(type(values).__name__)
print(values[1], len(values))
print(list(values))
print(list(values))
print(iter(values) is values)
```

range یک دنبالهٔ تکرارپذیر است و عناصر را به‌صورت یک فهرست کامل نگه نمی‌دارد. قابل اندیس‌گذاری و قابل پیمایش مجدد است. generator معمولی این دو ویژگی را ندارد. در برخی منابع قدیمی، range به‌طور غیر دقیق مولد نامیده شده است؛ در این کتاب، رفتار واقعی Python 3 ملاک تفکیک است.

## C08 — next با مقدار پیش‌فرض
گاهی نبودن عنصر بعدی یک پایان طبیعی است. آرگومان دوم next اجازه می‌دهد بدون تولید استثنا آن را تشخیص دهیم.

```python
source = iter([None, 7])
missing = object()
print(next(source, missing) is None)
print(next(source, missing))
print(next(source, missing) is missing)
```

از None به‌عنوان پایان استفاده نکردیم، چون در این داده یک مقدار واقعی است. object یک نشانهٔ هویتی خصوصی می‌سازد و مقایسه با is است. هر بار next وضعیت پیمایش را جلو می‌برد؛ درخواست آزمایشی هم بی‌اثر نیست. بدون آرگومان پیش‌فرض، پایان عادی با StopIteration اعلام می‌شود. حلقهٔ for این پایان را در سازوکار خود مدیریت می‌کند.

## C09 — پیمایش هم‌زمان با قرارداد طول
در تحلیل داده، کنار هم قرارگرفتن دو فهرست به معنای هم‌ترازی واقعی نیست. strict فقط بخشی از قرارداد را کنترل می‌کند.

```python
ids = ["B01", "B02"]
counts = [4, 6]
for position, (batch_id, count) in enumerate(
        zip(ids, counts, strict=True), start=1):
    print(position, batch_id, count)
```

zip و enumerate خودشان تکرارگرند؛ ساختنشان همهٔ داده را مصرف نمی‌کند. strict=True ناسازگاری طول را هنگام پیمایش تشخیص می‌دهد، نه لزوماً هنگام ساخت zip. این گزینه از Python 3.10 در دسترس است. برابری طول هرگز ثابت نمی‌کند که B01 به شمارش درست وصل شده است؛ برای اتصال معنایی معمولاً کلید مشترک لازم است.

## C10 — sorted یک مصرف‌کنندهٔ کامل است
زنجیرهٔ تنبل ممکن است به یک گام غیرتنبل برسد. sorted برای مرتب‌سازی کلی باید همهٔ ورودی محدود را مصرف کند.

```python
source = (value for value in [30, 10, 20])
ordered = sorted(source)
print(ordered)
print(list(source))
print(type(ordered).__name__)
```

## C11 — مصرف ناقص در any
مصرف‌کننده‌ها همیشه همهٔ ورودی را نمی‌خوانند. توقف زودهنگام هم مزیت دارد و هم بر وضعیت تکرارگر اثر می‌گذارد.

```python
source = iter([0, 0, 5, 8])
print(any(source))
print(list(source))
print(all([]), any([]))
```

any تا اولین مقدار truthy می‌رود و all تا اولین مقدار falsy. تهی بودن داده با «همه معتبرند» از نظر تجاری یکی نیست؛ all روی ورودی تهی True می‌دهد. بنابراین کنترل وجود رکورد را جدا بنویسید. مصرف ناقصِ یک مولد فایل‌خوان به مدیریت قطعی منبع هم نیاز دارد؛ به آن در مثال هجدهم برمی‌گردیم.

## C12 — map و filter تنبل‌اند
استفاده از یک ابزار تابعی فقط وقتی مفید است که خوانایی را بهتر کند. همین تبدیل را اغلب می‌توان با عبارت مولد خواناتر نوشت.

```python
values = [1, 2, 3, 4]
selected = filter(lambda value: value % 2 == 0, values)
scaled = map(lambda value: value * 10, selected)
print(next(scaled))
print(list(scaled))
print(list(scaled))
```

lambda اینجا تابع کوچکِ بی‌نام برای یک عبارت است؛ محل مناسبی برای منطق چندمرحله‌ای نیست. map و filter فهرست نمی‌سازند و در این نمونه تک‌مصرف‌اند. کوتاه‌نویسی معیار برتری نیست: اگر شرط یا تبدیل نیاز به نام و توضیح دارد، تابع معمولی بهتر است. قاعدهٔ مهم، روشن‌بودن مصرف و قرارداد ورودی است.

## C13 — عبارت مولد در برابر فهرست‌سازی
دو شکل ظاهراً مشابه در زمان انجام کار و امکان پیمایش دوباره تفاوت دارند.

```python
values = [2, 4, 6]
eager = [value * 2 for value in values]
lazy = (value * 2 for value in values)
print(type(eager).__name__, type(lazy).__name__)
print(sum(lazy), sum(lazy))
print(sum(eager), sum(eager))
```

فهرست خروجی محاسبه را اکنون نگه می‌دارد؛ مولد هر مقدار را هنگام درخواست تولید می‌کند. اما این مثال هنوز فهرست values را در حافظه دارد. جایگزینی براکت با پرانتز نه ورودی قبلی را حذف می‌کند و نه به‌خودی‌خود برنامه را سریع‌تر می‌کند. برای مقایسهٔ کارایی باید حجم داده، هزینهٔ تبدیل و الگوی مصرف را بسنجیم، نه تنها اندازهٔ خود شیء مولد را.

## C14 — تابع مولد و نقطهٔ تعلیق yield
برای فهم lazy بودن، زمان اجرا را قابل مشاهده می‌کنیم. ساخت شیء مولد با اجرای بدنهٔ تابع فرق دارد.

```python
def steps():
    print("started")
    yield 10
    print("resumed")
    yield 20

source = steps()
print("created")
print(next(source))
print(next(source))
```

وجود yield در بدنه، این تابع را generator function می‌کند. فراخوانی steps شیء generator می‌سازد؛ بدنه در نخستین درخواست شروع می‌شود. در yield وضعیت محلی حفظ و اجرای بدنه معلق می‌شود. پس از دومین مقدار هم پایان بدنه هنوز در next بعدی مشخص می‌شود. متغیرهای محلی و ارجاع‌های نگه‌داشته‌شده تا زمان مناسب می‌توانند زنده بمانند؛ تنبلی به معنی نداشتن وضعیت نیست.

## C15 — return در مولد یعنی پایان
مولد می‌تواند شرط توقف اختصاصی داشته باشد. yield مقدار بعدی را تحویل می‌دهد؛ return بدون مقدار پایان می‌دهد.

```python
def until_negative(values):
    for value in values:
        if value < 0:
            return
        yield value

print(list(until_negative([3, 0, -1, 9])))
print(list(until_negative([])))
```

صفر در این قرارداد یک دادهٔ عادی است و فقط مقدار منفی پایان می‌دهد. مقدار ۹ هرگز تولید نمی‌شود. این تصمیم باید معنای دامنه داشته باشد؛ استفادهٔ تصادفی از یک مقدار واقعی به‌عنوان علامت پایان خطرناک است. برای پایان طبیعی، درون مولد صریحاً StopIteration پرتاب نکنید؛ در Python جدید می‌تواند به RuntimeError تبدیل شود. return راه روشن این الگوست.

## C16 — yield from برای واگذاری پیمایش
اگر هدف تحویل‌دادن عناصر یک تکرارپذیر دیگر است، yield from یک الگوی مستقیم است؛ نه میان‌بری برای همهٔ توابع.

```python
def combined():
    yield "header"
    yield from [10, 20]
    yield from (value * 2 for value in [3, 4])

print(list(combined()))
```

در حد این فصل، yield from را تحویل عناصر منبعِ زیرین می‌خوانیم. به‌کاربردن yield به‌تنهایی روی یک فهرست، خود فهرست را یک عنصر تحویل می‌دهد؛ yield from عناصر آن را یکی‌یکی می‌دهد. سازوکارهای پیشرفتهٔ send و throw و coroutine در دامنهٔ فصل نیستند. یک منبع نامتناهی در میانهٔ این تابع مانع رسیدن به مراحل بعد می‌شود.

## C17 — تنبلی و تغییر مقدار بیرونی
تنبلی می‌تواند زمان مشاهدهٔ مقدار را عوض کند. این خطای منطقی رایج است، چون ساختن مولد از نظر ظاهری شبیه ثبت نتیجه به نظر می‌رسد.

```python
factor = 2
values = [1, 2]
source = (factor * value for value in values)
factor = 10
print(list(source))
```

ضرب در زمان مصرف انجام شده و factor آن زمان ۱۰ است. این نتیجه به معنی کپی‌شدن همهٔ متغیرها در زمان ساخت نیست. عبارت تکرارپذیرِ بیرونی یک generator expression هنگام ساخت ارزیابی می‌شود، ولی محاسبهٔ عنصر و شرط‌ها عموماً هنگام مصرف‌اند. اگر snapshot می‌خواهید باید مرز تثبیت مقدار را آگاهانه بسازید؛ صرف تابع‌کردن کد هم لزوماً کپی عمیق داده نمی‌سازد.

## C18 — مولد فایل‌خوان و بستن زودهنگام
فایل باید تا هنگام مصرف باز بماند. قرار دادن with داخل مولد طول عمر فایل را به طول عمر پیمایش مرتبط می‌کند.

```python
from contextlib import closing
from pathlib import Path
from tempfile import TemporaryDirectory

def iter_lines(path):
    with path.open(encoding="utf-8") as source:
        for line in source:
            yield line.rstrip("\n")

with TemporaryDirectory() as folder:
    path = Path(folder) / "values.txt"
    path.write_text("first\nsecond\n", encoding="utf-8")
    with closing(iter_lines(path)) as lines:
        print(next(lines))
    print(list(lines))
```

پس از خروج از closing، close مولد فراخوانی می‌شود و with داخلی فایل را می‌بندد. این الگو برای generator دارای close است، نه هر Iterable دلخواه. break به‌تنهایی تضمین عمومیِ بستن قطعی هر iterator نیست. بدون مصرف کامل، نگه‌داشتن ارجاع به مولد می‌تواند منبع را باز نگه دارد. خط نادرستِ رایج آن است که از درون with یک generator expression وابسته به فایل را return کنیم و سپس بیرون with بخوانیم؛ فایل پیش‌تر بسته شده است.

## C19 — Counter برای شمارش، نه جمع مبلغ
شمارش تعداد رکورد در هر دسته با جمع قیمت یا تعداد واحد کالا متفاوت است. نام شاخص را پیش از انتخاب ابزار مشخص کنید.

```python
from collections import Counter
categories = ["Books", "Digital", "Books", "Stationery"]
counts = Counter(categories)
print(sorted(counts.items()))
print(counts["Missing"])
print("Missing" in counts)
```

Counter بر اساس دفعات حضور همین عناصر ساخته شده است. مقدار صفر برای کلید غایب باعث ایجاد آن کلید نمی‌شود. شمار سفارش با تعداد کالا و مبلغ سفارش سه کمیت جدا هستند. برای خروجی پایدار، ترتیب را صریح تعیین کرده‌ایم. ترتیب تساوی‌ها در most_common از ترتیب نخستین مشاهده اثر می‌گیرد؛ اگر قرارداد تجاریِ مستقل می‌خواهید مرتب‌سازی چندمعیارهٔ خودتان را بنویسید.

## C20 — defaultdict و اثر خواندن کلید غایب
یک ابزار کمکی ممکن است خواندن ظاهری را به تغییر وضعیت تبدیل کند. دانستن این رفتار در اشکال‌زدایی مهم است.

```python
from collections import defaultdict
totals = defaultdict(int)
print(totals.get("Books"))
print("Books" in totals)
print(totals["Books"])
print("Books" in totals)
totals["Books"] += 120
print(dict(totals))
```

فراخوانی get از کارخانهٔ پیش‌فرض استفاده نمی‌کند، اما دسترسی با براکت به کلید غایب، مقدار می‌سازد و ثبت می‌کند. به defaultdict تابع int را می‌دهیم، نه نتیجهٔ int(). اگر کارخانه تعیین نشود، دسترسی به کلید غایب KeyError می‌دهد؛ ادعای برخی متن‌های قدیمی دربارهٔ مقدار خودکار None دقیق نیست. برای ثبت خروجی معمولاً dict معمولی مرز روشن‌تری می‌سازد.

## C21 — islice و chain: پیش‌نمایش با مصرف محدود
برای دیدن چند عنصر نخست، همهٔ جریان را به list تبدیل نمی‌کنیم. با این حال پیش‌نمایش هم ورودی را جلو می‌برد.

```python
from itertools import chain, count, islice
source = count(10, 2)
print(list(islice(source, 3)))
print(next(source))
print(list(chain([1, 2], [3], [])))
```

count اینجا منبع نامتناهی است و islice مصرف را محدود کرده است. list(count(...)) پایان نمی‌یابد و می‌تواند حافظه را تمام کند؛ آن را اجرا نکنید. chain عناصر منابع را به‌ترتیب تحویل می‌دهد و مرتب‌سازی یا ادغام بر اساس کلید انجام نمی‌دهد. islice نیز snapshot بی‌اثر نیست: منبع تا همان نقطه مصرف شده است. محدودکردن نمایش با محدودکردن دامنهٔ تحلیل یکی نیست.

## C22 — groupby فقط گروه‌های متوالی را می‌سازد
نام یک ابزار ممکن است شبیه ابزار دیگری در pandas یا SQL باشد، ولی قرارداد یکسانی نداشته باشد.

```python
from itertools import groupby
values = ["A", "A", "B", "A"]
for key, group in groupby(values):
    print(key, list(group))
```

دو گروه A داریم چون یک B بین آنها آمده است. این ابزار کل داده را برای پیدا کردن کلیدهای برابر جست‌وجو نمی‌کند. برای گروه‌بندی کلی می‌توان داده را با همان کلید مرتب کرد، یا در یک گذر از دیکشنری انباره استفاده کرد. مرتب‌سازی کامل هزینهٔ حافظه و زمان دارد؛ گروه‌بندی دیکشنری نیز به تعداد کلیدهای متمایز حافظه می‌خواهد. هیچ گزینه‌ای بدون دانستن سؤال برنده نیست.

## C23 — گروه‌بندی پس از مرتب‌سازی
اگر واقعاً همهٔ رکوردهای هم‌کلید را کنار هم می‌خواهیم، پیش‌شرط لازم را صریح ایجاد می‌کنیم.

```python
from itertools import groupby
from operator import itemgetter
rows = [("B", 3), ("A", 2), ("A", 5)]
key = itemgetter(0)
for category, group in groupby(sorted(rows, key=key), key=key):
    print(category, sum(row[1] for row in group))
```

itemgetter(0) یک callable برای برداشتن جزء اول می‌سازد؛ همان نقش lambda row: row[0] را دارد. گروه‌های groupby از منبع مشترک استفاده می‌کنند. گروه را پیش از جلو رفتن به گروه بعدی مصرف کنید؛ نگه‌داشتن iterator گروه برای مصرف بعدی می‌تواند به خروجی خالی برسد. اگر نیاز به نگهداری دارید باید دادهٔ گروه را آگاهانه materialize کنید و هزینه‌اش را بپذیرید.

## C24 — دسته‌بندی جریان با اندازهٔ محدود
برای تبدیل یا ارسال دسته‌ای، می‌توان فقط یک دستهٔ کوچک را نگه داشت. این نسخه برای Python 3.11 نوشته شده و به batched نسخه‌های جدید متکی نیست.

```python
from itertools import islice

def chunks(values, size):
    if type(size) is not int or size <= 0:
        raise ValueError("size must be a positive integer")
    source = iter(values)
    while True:
        batch = tuple(islice(source, size))
        if not batch:
            return
        yield batch

print(list(chunks(range(7), 3)))
```

دستهٔ آخر ممکن است کوتاه باشد. بررسی size داخل بدنهٔ مولد است و تا نخستین مصرف اجرا نمی‌شود؛ اگر اعتبارسنجی فوری لازم است باید پوستهٔ معمولی بنویسیم که مولد داخلی را برگرداند. خود chunks یک دسته را نگه می‌دارد، اما list دور آن در این نمایش کوچک همهٔ دسته‌ها را جمع می‌کند. در مصرف واقعی برای هر دسته داخل حلقه پردازش کنید. این تابع خودش منبع ورودی را close نمی‌کند.

## C25 — کلاس، نمونه و ویژگی
هدفِ شی‌گرایی در این کتاب فهم API و مدل‌کردن وضعیت لازم است، نه تبدیل هر دیکشنری به یک کلاس.

```python
class Batch:
    def __init__(self, batch_id, tested):
        self.batch_id = batch_id
        self.tested = tested

first = Batch("B01", 80)
second = Batch("B02", 120)
print(first.batch_id, first.tested)
print(second.batch_id, second.tested)
print(first is second)
```

Batch کلاس است و first و second دو نمونهٔ آن‌اند. self پارامتر قراردادیِ اشاره به نمونه است، نه واژهٔ رزروشده. __init__ نمونهٔ ساخته‌شده را مقداردهی اولیه می‌کند و نباید یک مقدار دلخواه برگرداند. این کلاس هنوز هیچ اعتبارسنجی ندارد: اسم فیلد و شکل سازنده دلیل معتبر بودن tested نیست. اگر تنها هدف حمل دادهٔ کوچک است، دیکشنری یا dataclass هم ممکن است کافی باشد.

## C26 — متد و اتصال نمونه
متد معمولی یک تابع در تعریف کلاس است که هنگام دسترسی از نمونه، آن نمونه را برای پارامتر نخست دریافت می‌کند.

```python
class Batch:
    def __init__(self, tested, rejected):
        self.tested = tested
        self.rejected = rejected

    def rejection_rate(self):
        if self.tested == 0:
            return None
        return self.rejected / self.tested

batch = Batch(80, 4)
print(batch.rejection_rate())
print(Batch.rejection_rate(batch))
print(Batch(0, 0).rejection_rate())
```

دو فراخوانی نخست در این مثال نتیجهٔ یکسان دارند؛ شکل نخست معمول‌تر است. این متد محاسبه می‌کند و چاپ نمی‌کند. None در مخرج صفر یعنی نرخ در این قرارداد تعریف نشده، نه نرخ صفر. منفی نبودن و رابطهٔ rejected و tested هنوز کنترل نشده‌اند؛ این کلاس برای دادهٔ ازپیش‌معتبر است. راهنمای نوع هم اگر افزوده شود، به‌تنهایی این کنترل را انجام نمی‌دهد.

## C27 — ظرف تغییرپذیر باید متعلق به نمونه باشد
یکی از دام‌های کلاس، اشتراک ناخواستهٔ فهرست در سطح کلاس است. نسخهٔ صحیح در __init__ ظرف تازه می‌سازد.

```python
class Batch:
    def __init__(self, batch_id):
        self.batch_id = batch_id
        self.notes = []

first = Batch("B01")
second = Batch("B02")
first.notes.append("checked")
print(first.notes)
print(second.notes)
print(first.notes is second.notes)
```

اگر notes = [] را در بدنهٔ کلاس و بیرون __init__ بگذاریم، نمونه‌هایی که ویژگی جدا ندارند همان فهرست کلاسی را می‌بینند. هر بار ساخت نمونه در این نسخه، یک فهرست مستقل می‌سازد. این اصل شبیه دام پارامتر پیش‌فرض تغییرپذیر فصل هشتم است، اما محل و زمان ساخت شیء فرق دارد. صفت کلاسی برای ثابت یا وضعیت عمداً مشترک می‌تواند درست باشد؛ مسئله اشتراک ناخواسته است.

## C28 — نام دوم، نمونهٔ دوم نیست
قواعد ارجاع فصل هفتم در مورد نمونه‌های کلاس هم برقرارند. کلاس، انتساب را به کپی تبدیل نمی‌کند.

```python
class Reading:
    def __init__(self, value):
        self.value = value

original = Reading(12)
alias = original
independent = Reading(original.value)
alias.value = 15
print(original.value, independent.value)
print(alias is original)
```

alias و original دو نام برای یک شیء هستند. independent یک نمونهٔ تازه با مقدار اولیهٔ مشابه است؛ در این مثال فیلد فقط عدد است. اگر فیلد یک فهرست بود، ساخت نمونهٔ تازه با همان فهرست الزاماً دادهٔ داخلی را مستقل نمی‌کرد. کپی سطحی و عمیق همان تفاوت‌های قبلی را دارند. بنابراین واژهٔ «نسخهٔ مستقل» باید سطح استقلال را مشخص کند.

## C29 — dataclass برای دادهٔ ساختاریافتهٔ کوچک
مسیر تکمیلی: یک decorator آماده می‌تواند بخشی از کد تکراری سازنده و نمایش را تولید کند. لازم نیست اکنون سازوکار نوشتن decorator را بیاموزیم.

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class Amount:
    gross: int
    refund: int

    def retained(self):
        return self.gross - self.refund

amount = Amount(500, 120)
print(amount)
print(amount.retained())
print(amount == Amount(500, 120))
```

dataclass به‌طور معمول __init__ و نمایش و برابری مبتنی بر فیلدها را می‌سازد. frozen جلوی انتساب معمول به فیلد را می‌گیرد، نه هر تغییر عمیق در هر شیء داخلی و نه هر دستکاری عمدی را. راهنمای int در اینجا مقدار منفی یا ورودی رشته‌ای را خودکار رد نمی‌کند. برای فهرست پیش‌فرض در dataclass باید default_factory مناسب داشت؛ از این ابزار به‌عنوان جایگزین قرارداد معتبر استفاده نکنید.

## C30 — دو شاخص، یک گذر
پیش از پروژهٔ کامل، خط اصلی طراحی را در یک تابع کوچک می‌بینیم: تعداد و جمع را از همان پیمایش به دست آوریم.

```python
def summarize(values):
    count = 0
    total = 0
    for value in values:
        count += 1
        total += value
    average = total / count if count else None
    return count, total, average

source = (value for value in [10, 20, 30])
print(summarize(source))
print(summarize(source))
print(summarize([]))
```

این تابع ورودی را دقیقاً یک بار می‌پیماید. دوباره‌دادن همان مولدِ مصرف‌شده، مجموعهٔ تهی می‌دهد. count و total وضعیت کوچکِ کافی برای این دو شاخص‌اند، اما برای میانهٔ دقیقِ عمومی یا تشخیص یکتایی همهٔ شناسه‌ها وضعیت بیشتری لازم است. نام «جریانی» باید به الگوریتم و کل مسیر مربوط باشد، نه فقط به وجود yield در یک فایل.
