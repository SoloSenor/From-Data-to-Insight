## محاسبهٔ آزمایشی

این مثال از فصل اول است: ۲۷۰۰ سفارش در ۲۷ روز کاری.
خروجی، فقط صحت اجرای یک محاسبهٔ کوچک را می‌سنجد.

```python
orders = 2700
working_days = 27
orders_per_day = orders / working_days
print('Orders per working day:', orders_per_day)
```

```python
assert sys.prefix != sys.base_prefix, 'Select the project venv kernel.'
assert working_days > 0
assert orders_per_day == 100.0
print('Basic checks passed')
```
