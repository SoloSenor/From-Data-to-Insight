# فصل هفتم: ساختارهای داده و الگوهای پردازش
سلول‌ها را از ابتدا به ترتیب اجرا کنید؛ کد و خروجی انگلیسی است. خطاهای عمدی در این دفترچه نیستند.

## 01. دو نام، یک شیء

انتساب، شیء را تکثیر نمی‌کند. دو نام می‌توانند به همان فهرست اشاره کنند. پیش از اجرا حدس بزنید تغییر از مسیر نام دوم در نام اول دیده می‌شود یا نه.

هر دو نام به یک فهرست اشاره می‌کنند. عمل append محتوای همان شیء را تغییر داده است. عملگر is یکسان‌بودن شیء را بررسی می‌کند؛ برای مقایسهٔ مقدارهای معمول از == استفاده می‌کنیم. به‌ویژه برای عدد و رشته نباید بر رفتار حافظه یا ذخیره‌سازی داخلی تکیه کرد.

```python
values = [10, 20]
alias = values
alias.append(30)
print(values)
print(alias is values)
```

## 02. برابری مقدار با هویت یکی نیست

دو ظرف مستقل می‌توانند فعلاً محتوای برابر داشته باشند.

برابری اولیه تضمین نمی‌کند دو نام یک شیء باشند. پس از تغییر فهرست دوم، فهرست اول دست‌نخورده است. برای تشخیص None استفاده از is None مناسب است؛ اما این توصیه به معنی استفاده از is برای مقایسهٔ همهٔ مقدارها نیست.

```python
left = [10, 20]
right = [10, 20]
print(left == right)
print(left is right)
right.append(30)
print(left)
```

## 03. کپی کم‌عمق: لایهٔ بیرونی مستقل

کپی یک فهرست تخت از عددها برای مستقل‌کردن تغییر جایگاه‌ها کافی است.

فهرست بیرونی تازه است. جایگزینی عضو اول کپی، جایگاه اول فهرست اصلی را عوض نمی‌کند. عددها خود تغییرپذیر نیستند؛ این مثال هنوز دربارهٔ ظرف‌های داخلی چیزی ثابت نمی‌کند.

```python
original = [10, 20]
shallow = original.copy()
shallow[0] = 99
print(original)
print(shallow)
print(original is shallow)
```

## 04. کپی بیرونی، اشتراک درونی

این بار عضو فهرست یک دیکشنری است. کپی سطح اول، دیکشنری داخلی را کپی نمی‌کند.

عمل append به فهرست داخلی رسیده است؛ هر دو مسیر همان داده را می‌بینند. اشتراک ساختار لزوماً خطا نیست، اما باید آگاهانه باشد. همین مسئله برای dict.copy و برش کامل فهرست نیز وجود دارد.

```python
source = [{"order_id": "A", "tags": ["new"]}]
shallow = source.copy()
shallow[0]["tags"].append("checked")
print(source[0]["tags"])
print(shallow is source)
print(shallow[0] is source[0])
```

## 05. کپی هر رکورد هم همیشه کافی نیست

در یک فهرست رکورد، کپی‌کردن دیکشنری هر سطر سطح دیگری را مستقل می‌کند، نه همهٔ عمق را.

شناسهٔ اصلی A مانده، اما برچسب داخلی تغییر کرده است. درک فهرستی را فعلاً به معنای «برای هر سطر، یک کپی بساز» بخوانید؛ ساختار آن را کمی بعد باز می‌کنیم. استقلال باید نسبت به نوع تغییر موردنظر تعریف شود.

```python
source = [{"order_id": "A", "tags": ["new"]}]
copied_rows = [row.copy() for row in source]
copied_rows[0]["order_id"] = "B"
copied_rows[0]["tags"].append("checked")
print(source[0]["order_id"])
print(source[0]["tags"])
print(copied_rows[0] is source[0])
```

## 06. کپی عمیق برای ظرف‌های داخلی معمول

برای این دادهٔ کوچکِ متشکل از فهرست و دیکشنری و مقدارهای ساده، deepcopy نسخه‌ای مستقل برای تغییر داخلی می‌سازد.

واردکردن deepcopy فقط دسترسی به ابزار کتابخانهٔ استاندارد است؛ سازمان‌دهی ماژول‌ها موضوع فصل هشتم است. کپی عمیق راه‌حل همیشگی نیست: زمان و حافظه می‌خواهد، برای منابعی مانند فایل باز نسخهٔ مستقلِ قابل‌استفاده تضمین نمی‌کند و ممکن است اشتراک عمدیِ زیرشیءها را درون کپی حفظ کند. ابتدا تعیین کنید چه چیزی باید مستقل شود.

```python
from copy import deepcopy
source = [{"order_id": "A", "tags": ["new"]}]
independent = deepcopy(source)
independent[0]["tags"].append("checked")
print(source[0]["tags"])
print(independent[0]["tags"])
```

## 07. دام تکرار ظرف تغییرپذیر

ضرب فهرست، عضو داخلی را سه بار کپی عمیق نمی‌کند؛ ارجاع همان عضو را تکرار می‌کند.

در نسخهٔ دوم، عبارت ساخت فهرست در هر تکرار دوباره اجرا می‌شود. زیرخط فقط نامی برای متغیری است که به مقدارش نیاز نداریم؛ دستور ویژه نیست. همین دام در ساخت چند سطل گروه‌بندی با یک فهرست مشترک نیز رخ می‌دهد.

```python
bad_grid = [[0, 0]] * 3
bad_grid[0][1] = 7
print(bad_grid)
good_grid = [[0, 0] for _ in range(3)]
good_grid[0][1] = 7
print(good_grid)
```

## 08. ساختار ثابت، عضو تغییرپذیر

تغییرناپذیری تاپل به معنی ثابت‌بودن همهٔ اشیای داخل آن نیست.

جایگاه‌های تاپل عوض نشده‌اند؛ عضو دوم همچنان همان فهرست است، اما محتوایش تغییر کرده است. جایگزینی مستقیم یک جایگاه تاپل خطاست؛ نمونهٔ خطا در بخش جداگانه آمده است.

```python
record = ("A", ["new"])
record[1].append("checked")
print(record)
```

## 12. جفت‌کردن دنباله‌ها با zip

zip عضوهای هم‌موقعیت را کنار هم قرار می‌دهد. برابری طول دو فهرست کافی نیست؛ ترتیب معنایی آنها هم باید هم‌تراز باشد.

strict=True در Python 3.10 به بعد موجود است. در این کتاب روی Python 3.11 آزموده شده است. اگر طول‌ها متفاوت باشند، هنگام مصرف زوج‌ها ValueError ایجاد می‌شود؛ نه لزوماً در لحظهٔ ساخت zip. این کنترل نمی‌فهمد که مثلاً فهرست دوم به‌اشتباه مستقل مرتب شده است.

```python
codes = ["A", "B", "C"]
quantities = [2, 5, 1]
for code, quantity in zip(codes, quantities, strict=True):
    print(code, quantity)
```

## 13. zip معمولی کوتاه می‌آید

سطر دوم خالی است چون همین iterator قبلاً مصرف شده است. برای پیمایش دوباره، zip را دوباره بسازید یا اگر داده کوچک است نتیجه را در فهرست نگه دارید. zip و enumerate همهٔ نتیجه را از پیش به صورت فهرست نمی‌سازند. strict فقط وقتی کنترل می‌شود که مصرف به محل اختلاف برسد؛ توقف زودهنگام می‌تواند اختلاف انتهایی را پنهان کند.

```python
pairs = zip(["A", "B", "C"], [2, 5])
print(list(pairs))
print(list(pairs))
```

## 14. از حلقهٔ روشن به درک فهرستی

ابتدا تبدیل را با حلقه می‌نویسیم، سپس همان معنا را به شکل درک فهرستی بیان می‌کنیم.

فرمول ذهنی این است: «چه بسازم، از کدام عضوها، با چه شرطی؟». عبارت ابتدای درک، عضو خروجی را می‌سازد و بخش for منبع را معرفی می‌کند. درک فهرستی فوراً یک فهرست می‌سازد و برای دادهٔ بسیار بزرگ ممکن است حافظهٔ زیادی بخواهد؛ کوتاه‌تر بودن تضمین سریع‌تر بودن نیست.

```python
values = [2, 5, 1]
scaled_loop = []
for value in values:
    scaled_loop.append(value * 10)
scaled_comp = [value * 10 for value in values]
print(scaled_loop)
print(scaled_loop == scaled_comp)
```

شرط انتهای درک عضو را حذف می‌کند؛ عبارت شرطی در ابتدای آن یک مقدار انتخاب می‌کند. kept در این برچسب‌گذاری فقط یعنی «هیچ واحد مرجوعی ثبت نشده»، نه تأیید کیفیت، وصول پول یا سودآوری. عبارت‌های شرطی تو‌در‌توی طولانی را به شرط و تابع روشن تبدیل کنید.

## 18. بازنویسی بی‌صدای کلید تکراری

این نمونه عمداً از شناسهٔ تکراری استفاده می‌کند؛ کد اجرا می‌شود ولی یک رکورد در نمایه باقی نمی‌ماند.

آخرین مقدار کلید A جای مقدار قبلی را گرفته است. این رفتار خطای اجرایی نیست؛ اگر قرارداد داده یکتایی شناسه باشد، نقض قرارداد است. مقایسهٔ طول نمایه و ورودی یا کنترل صریح پیش از انتساب می‌تواند آن را آشکار کند. حذف تکراری‌ها نیازمند تصمیم مستند است؛ نباید خودکار آخرین رکورد را حقیقت بدانیم.

```python
duplicate_rows = [
    {"order_id": "A", "quantity": 1},
    {"order_id": "A", "quantity": 9},
]
unsafe_index = {row["order_id"]: row for row in duplicate_rows}
print(len(duplicate_rows), len(unsafe_index))
print(unsafe_index["A"]["quantity"])
```

## 19. ساخت نمایه با کنترل یکتایی

در این تابع، برای شناسهٔ تکراری تصمیم صریح می‌گیریم: توقف با ValueError، نه بازنویسی پنهان.

raise خطا را آگاهانه ایجاد می‌کند؛ مدیریت استثنا و traceback در فصل نهم مفصل می‌آید. این تابع فقط تکرار کلید را کنترل می‌کند، نه کامل‌بودن schema و معتبر بودن قیمت و تعدادها. همچنین نمایه همچنان ارجاع‌های رکوردها را نگه می‌دارد. ورودی این تابع دنباله‌ای از رکوردهای معتبر با شناسهٔ رشته‌ای است.

```python
def index_unique(rows):
    index = {}
    for row in rows:
        order_id = row["order_id"]
        if order_id in index:
            raise ValueError("Duplicate order_id")
        index[order_id] = row
    return index

checked_index = index_unique(orders)
print(len(checked_index))
print(index_unique([]))
```

## 20. مرتب‌سازی با تابع کلید

مرتب‌کردن رکوردها را بر اساس یک ویژگی تعریف می‌کنیم. تابع کلید برای هر رکورد مقدار قابل‌مقایسه را برمی‌گرداند.

ورودی تغییر ترتیب نداده است. برای تعدادهای برابر، مرتب‌سازی پایدار ترتیب قبلی را حفظ می‌کند؛ بنابراین O006 پیش از O007 مانده است. sorted فهرست تازه می‌سازد اما رکوردهای داخلی را عمیق کپی نمی‌کند. reverse جهت ترتیب را عوض می‌کند، نه معنای شاخص را.

```python
def quantity_key(row):
    return row["quantity"]

ranked_quantity = sorted(orders, key=quantity_key, reverse=True)
print([row["order_id"] for row in ranked_quantity[:4]])
print(orders[0]["order_id"])
```

## 21. lambda فقط یک تابع کوچک است

اگر تابع کلید بسیار ساده است، می‌توان آن را با lambda نوشت. برای منطق چندمرحله‌ای همان def روشن‌تر است.

در lambda نام row ورودی است و عبارت پس از دونقطه مقدار برگشتی را می‌سازد. این تابع نام مستقل ندارد. خروجی دوم یادآوری می‌کند مرتب‌سازی، ظرف تازه‌ای از ارجاع رکوردها می‌سازد؛ تغییر رکورد خروجی می‌تواند دادهٔ اصلی را عوض کند.

```python
ranked_lambda = sorted(
    orders, key=lambda row: row["quantity"], reverse=True
)
print([row["order_id"] for row in ranked_lambda[:4]])
print(ranked_lambda[0] is orders[2])
```

## 22. کلید چندجزئی و شکستن تساوی

برای مرتب‌سازی تعداد نزولی و شناسهٔ صعودی، کلید را یک تاپل می‌گیریم. این بار ترتیب معکوس ورودی نباید ترتیب شناسه‌های هم‌تعداد را تعیین کند.

تاپل‌ها جزءبه‌جزء مقایسه می‌شوند: اول تعداد منفی، سپس شناسه. منفی‌کردن اینجا مخصوص مقدار عددی است و راه عمومی برای نزولی‌کردن متن نیست. reverse=True روی کل تاپل جهت همهٔ مؤلفه‌ها را معکوس می‌کند؛ برای جهت‌های متفاوت باید کلید یا چند مرتب‌سازی پایدار را درست طراحی کرد.

```python
ranked_ties = sorted(
    orders[::-1],
    key=lambda row: (-row["quantity"], row["order_id"]),
)
print([row["order_id"] for row in ranked_ties[:6]])
```

## 24. غنی‌سازی رکورد بدون تغییر ورودی

برای هر سفارش یک دیکشنری تازه می‌سازیم و مبلغ اولیه، بازپرداخت و مبلغ پس از مرجوعی را به آن اضافه می‌کنیم.

چون فیلدهای موجود این داده همه مقدارهای ساده‌اند و فقط فیلدهای سطح اول اضافه می‌کنیم، کپی هر دیکشنری برای این کار کافی است. این ادعا به همهٔ داده‌های تو‌در‌تو تعمیم ندارد. مبلغ پس از مرجوعی سود نیست؛ هزینه، مالیات و تخفیف در این سناریو نداریم. اگر بعداً تعداد یا قیمت عوض شود، مبلغ ذخیره‌شده خودکار دوباره محاسبه نمی‌شود.

```python
enriched = []
for row in orders:
    item = row.copy()
    gross = row["quantity"] * row["unit_price_toman"]
    refund = row["returned_quantity"] * row["unit_price_toman"]
    item["gross_toman"] = gross
    item["refund_toman"] = refund
    item["retained_toman"] = gross - refund
    enriched.append(item)
print(enriched[0]["retained_toman"])
print("retained_toman" in orders[0])
```

## 25. تجمیع خوانا با انبارهٔ دیکشنری

برای هر گروه یک مبلغ انباشته نگه می‌داریم. اینجا get با صفر مجاز است چون گروهِ هنوز دیده‌نشده در جمع ما از صفر شروع می‌کند، نه اینکه دادهٔ نامعلوم را صفر فرض کنیم.

کلیدهای خروجی برای نمایش قطعی مرتب شده‌اند. این کار جمع قیمت واحد نیست؛ مقدار کالا و مرجوعی قبلاً در مبلغ هر سفارش لحاظ شده‌اند. برای محاسبات چندمرحله‌ای، حلقه از یک درک پیچیده و دارای اثر جانبی روشن‌تر است. خروجی اینجا فقط گروه‌های مشاهده‌شده را دارد.

```python
totals = {}
for row in enriched:
    category = row["category"]
    totals[category] = totals.get(category, 0) + row["retained_toman"]
for category, amount in sorted(totals.items()):
    print(category, amount)
```

## 26. سهم گروه‌ها با مخرج روشن

سهم اینجا نسبت مبلغ پس از مرجوعی هر گروه به جمع همان مبلغ در همهٔ گروه‌هاست.

اگر جمع مبلغ صفر باشد، در این نسخهٔ نمایشی دیکشنری سهم خالی است: سهم تعریف نشده و خروجی باید همین قرارداد را توضیح دهد، نه سهم صفر جعلی بسازد. در کارگاه تابع‌محور، برای هر گروه سهم None ثبت می‌کنیم تا نام گروه‌ها از بین نرود. این سهم، سهم تعداد سفارش یا سهم تعداد واحد نیست. قالب‌بندی دو رقم اعشار فقط نمایش را گرد می‌کند.

```python
grand_total = sum(totals.values())
shares = {
    category: amount / grand_total
    for category, amount in totals.items()
} if grand_total else {}
print(grand_total)
for category in sorted(shares):
    print(category, f"{shares[category]:.2%}")
```

## 27. گروه‌بندی رکوردها با ظرف مستقل

گاهی به جای فقط جمع، خود رکوردهای هر گروه لازم‌اند. برای هر گروه در اولین برخورد یک فهرست تازه بسازید.

فهرست گروه‌ها مستقل‌اند اما رکوردهای داخلشان همان رکوردهای اصلی هستند. dict.fromkeys با یک فهرست خالی مشترک انتخاب مناسبی برای این کار نیست. شمارش چهار سفارش در هر گروه، به معنی برابری تعداد واحد یا مبلغ نیست.

```python
groups = {}
for row in orders:
    category = row["category"]
    if category not in groups:
        groups[category] = []
    groups[category].append(row)
print([(name, len(groups[name])) for name in sorted(groups)])
print(groups["Books"] is groups["Digital"])
```

## 28. تخت‌کردن یک لایه با حلقهٔ روشن

برای نمونه‌ای از سفارش چندقلمی، فهرست قلم‌ها را به ردیف‌های قابل‌بررسی تبدیل می‌کنیم. این داده از سفارش‌های تک‌قلمی کارگاه جداست.

این خروجی دو قلم دارد، اما ورودی دو سفارش دارد و سفارش B بدون قلم در خروجی دیده نمی‌شود. پس از تخت‌کردن، واحد مشاهده عوض شده است؛ len خروجی را تعداد سفارش نخوانید. یک درک با دو for هم ممکن است، اما حلقهٔ تو‌در‌توی کوتاه برای آغاز خواناتر است. این کار فقط همین یک لایه را تخت می‌کند.

```python
baskets = [
    {"order_id": "A", "items": [{"sku": "P1"}, {"sku": "P2"}]},
    {"order_id": "B", "items": []},
]
lines = []
for basket in baskets:
    for item in basket["items"]:
        lines.append((basket["order_id"], item["sku"]))
print(lines)
```

## 30. گزارش نهایی و کنترل سازگاری

اکنون گزارش گروهی را بر اساس مبلغ نزولی رتبه‌بندی می‌کنیم. مرتب‌سازی دیکشنری با items زوج‌های گروه و مبلغ را در کنار هم نگه می‌دارد.

در sum، عبارت داخل پرانتز یک عبارت مولد است: مقدارها را برای مصرف یکی‌یکی می‌دهد و فهرست میانی نمی‌سازد. معرفی کامل مولدها خارج از دامنهٔ این فصل است؛ فعلاً این الگوی سادهٔ جمع را بشناسید. درست‌بودن تراز مالی، به‌تنهایی قیمت و تعداد ورودی را تأیید نمی‌کند. نتیجه با فصل ششم یکسان است، اما این بار مسیر محاسبه فقط بر ساختارهای پایهٔ پایتون تکیه دارد.

```python
gross_total = sum(row["gross_toman"] for row in enriched)
refund_total = sum(row["refund_toman"] for row in enriched)
retained_total = sum(row["retained_toman"] for row in enriched)
ranking = sorted(totals.items(), key=lambda pair: (-pair[1], pair[0]))
print("gross_toman", gross_total)
print("refund_toman", refund_total)
print("retained_toman", retained_total)
print("balanced", gross_total - refund_total == retained_total)
for rank, (category, amount) in enumerate(ranking, start=1):
    print(rank, category, amount)
```
